## Strings  (indexes, escape sequences)


```python

s = "Python!"
print("First char:", s[0])
print("Second char:", s[1])
print("Last char:", s[-1])   # negative index counts from end

# Escape sequences
print("Line1\nLine2")  # \n = newline
print("Quote: \"Hello\"")  # \" inside a string
```

## Formatted strings
Use `f"...{var}..."` (f-strings) for readable formatting.


```python

name = "Ellis"
followers = 1280
print(f"{name} has {followers} followers.")
print(f"{name} has {followers/1000:.1f}K followers.")
```

## List basics
Lists are *mutable* containers.


```python

nums = [10, 20, 30]
nums.append(40)      # add to end
nums[1] = 99         # modify
print(nums)

# Iterate
for n in nums:
    print("n:", n)
```

## Tuple basics
Tuples are *immutable* containers.


```python

point = (3, 4)
x, y = point  # unpacking
print("x:", x, "y:", y)

# point[0] = 99  # TypeError: tuples cannot be modified
```
