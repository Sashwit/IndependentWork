```python
#Exercise 4
#a) Mean = 14.857
#b) Variance = Σx² = 100 + 144 + 196 + 196 + 256 + 324 + 400 = 1616, Σ(x − x̄)² = Σx² − (Σx)²/n = 1616 − 104²/7 = 1616 − 1545.14 ≈ 70.86
#c) Standard Deviation = Population: σ = √10.12 ≈ 3.18.

import numpy as np

data = [10, 12, 14, 14, 16, 18, 20]

# Step 1: Mean
mean = sum(data) / len(data)
print(f"Step 1 - Mean: {mean}")  # = 5

# Steps 2 & 3: Subtract mean, then square
deviations_squared = [(x - mean)**2 for x in data]
print(f"Step 2 & 3 - Squared deviations: {deviations_squared}")

# Step 4: Average of squared deviations = Variance
variance = sum(deviations_squared) / len(data)
print(f"Step 4 - Variance: {variance}")  # = 4

# Standard Deviation = Square Root of Variance
std_dev = variance ** 0.5
print(f"Standard Deviation: {std_dev}")  # = 2

# Verify with numpy
print(f"\nVerify with numpy: var = {np.var(data)}, std = {np.std(data)}")
```

```python
#Exercise 5
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats # imports SciPy's statistics module

# Visualise the Normal Distribution (Bell Curve)
mu, sigma = 70, 10  # UK adult male heights
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 300)
y = stats.norm.pdf(x, mu, sigma)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, y, 'b-', linewidth=2)

# Shade the regions
colors = ['#d4e6f1', '#a9cce3', '#7fb3d3']
labels = ['68% (μ ± 1σ)', '95% (μ ± 2σ)', '99.7% (μ ± 3σ)']
for i, (n, col, lab) in enumerate(zip([1, 2, 3], colors, labels)):
    ax.fill_between(x, y, where=(x >= mu - n*sigma) & (x <= mu + n*sigma),
                    alpha=0.5, color=col, label=lab)

ax.set_title('Normal Distribution: Exam scores (μ = 70, σ = 10)', fontsize=13)
ax.set_xlabel('Exam scores')
ax.set_ylabel('Probability Density')
ax.legend()
ax.axvline(mu, color='red', linestyle='--', label='Mean')
plt.tight_layout()
plt.show()
```

(a) Between 60 and 80: about 68%:
60 and 80 are each one standard deviation from the mean (70 - 10 and 70 + 10). About 68% of values fall within ±1σ.

(b) Above 90: (about 2.5%) -
90 is two standard deviations above the mean (70 + 2 x 10). About 95% of values fall within ±2σ, which leaves 5% in the two tails. By symmetry, half of that is in the upper tail: 5% ÷ 2 = 2.5%.

(c) A score of 45: 2.5 standard deviations below the mean
