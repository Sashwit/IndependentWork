Type conversion
`int()`, `float()`, `str()` convert between common types.


```python
#Type conversion

s = "12"
n = int(s)      # "12" -> 12
f = float(s)    # "12" -> 12.0

print(n, type(n))
print(f, type(f))

# Converting to string for printing or joining
print("n as text: " + str(n))
```

Mixed data types
Be careful combining strings and numbers.


```python
#Mixed types

apples = 5
# print("I have " + apples + " apples")  # TypeError
print("I have " + str(apples) + " apples")  # OK

# Better: f-strings (also covered later)
print(f"I have {apples} apples")
```

#Floating-point errors
Some decimals cannot be represented exactly in binary.


```python
#Floating point precision

a = 0.1 + 0.2
print("0.1 + 0.2 =", a)
print("Is it exactly 0.3?", a == 0.3)

# When comparing floats, use a tolerance
tolerance = 1e-9
print("Close enough?", abs(a - 0.3) < tolerance)
```

## math module
The `math` module includes constants and functions like sqrt, sin, floor, ceil.


```python

import math

print("pi:", math.pi)
print("sqrt(81):", math.sqrt(81))
print("ceil(2.1):", math.ceil(2.1))
print("floor(2.9):", math.floor(2.9))
```
