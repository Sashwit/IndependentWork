## Exercise 1: Build Your Own Model

Use this dataset (hours studied vs exam score):
```
Hours: [1, 2, 3, 4, 5, 6, 7, 8]
Score: [35, 45, 55, 62, 70, 78, 85, 92]
```
(a) Set up the feature matrix X and target vector y.  
(b) Run gradient descent for 100 iterations to find w and bias.  
(c) Predict the score for 9 hours of study.  
(d) Calculate R².

````python
import numpy as np
#STEP A: Feature Matrix (X) and Target Vector (y)

# Feature matrix X — what we know (house sizes)
X = np.array([1, 2, 3, 4, 5, 6, 7, 8])  #8 hours for 8 scores

# Target vector y — what we want to predict (prices)
y = np.array([35, 45, 55, 62, 70, 78, 85, 92])

print('Feature Matrix X (Hours):')
print(f'  X = {X}')
print(f'  Shape: {X.shape} — {len(X)} hours')

print('Target Vector y (Scores):')
print(f'  y = {y}')
print(f'  Shape: {y.shape} — {len(y)} scores')

print('X hour(s) for y score(s):')
for i in range(len(X)):
    print(f'  House {i+1}: {X[i]:,} hour → {y[i]:,}')
````

````python
#Part B
import numpy as np

# ====== STEP 0: Scale the Data ======
# X values are 1-8, y values are 35-92
# Scaling X by 10 and y by 100 to get values roughly between 0 and 1
X_scaled = X / 10.0
y_scaled = y / 100.0

print('Scaled Data:')
print(f'  X_scaled = {X_scaled}')
print(f'  y_scaled = {y_scaled}\n')

# ====== STEP 1: Initialise Parameters ======
w = 0.0        # Initial weight
bias = 0.0     # Initial bias
lr = 0.1       # Learning rate
n = len(X_scaled) # Number of samples
epochs = 100   # Number of training iterations

# Store history for plotting (optional, but good for analysis)
history = {'epoch': [], 'w': [], 'bias': [], 'mse': []}

print('Training Linear Regression from scratch...')
print(f'{"Epoch":>6} | {"w":>10} | {"bias":>10} | {"MSE":>14}')
print('-' * 50)

# ====== STEP 2: Training Loop (Gradient Descent) ======
for epoch in range(epochs):
    # Forward pass: predict y_hat
    y_pred = w * X_scaled + bias

    # Compute loss (Mean Squared Error - MSE)
    mse = np.mean((y_pred - y_scaled) ** 2)

    # Store history
    history['epoch'].append(epoch)
    history['w'].append(w)
    history['bias'].append(bias)
    history['mse'].append(mse)

    # Print selected epochs to see progress
    if epoch % 10 == 0 or epoch == epochs - 1:
        print(f'{epoch:>6} | {w:>10.4f} | {bias:>10.4f} | {mse:>14.6f}')

    # Compute gradients (partial derivatives of MSE wrt w and bias)
    dw = (2/n) * np.sum((y_pred - y_scaled) * X_scaled)
    db = (2/n) * np.sum(y_pred - y_scaled)

    # Update parameters using gradient descent
    w = w - lr * dw
    bias = bias - lr * db

print('-' * 50)

print(f'\nFinal parameters after {epochs} epochs:')
print(f'  Scaled w = {w:.6f}')
print(f'  Scaled bias = {bias:.6f}')

# Unscaling the parameters to interpret them in the original units
# Original relationship: Y = w_orig * X_orig + bias_orig
# Scaled relationship: Y_scaled = w_scaled * X_scaled + bias_scaled
# (Y_orig / 100) = w_scaled * (X_orig / 10) + bias_scaled
# Y_orig = (w_scaled * 100 / 10) * X_orig + (bias_scaled * 100)
# Y_orig = (w_scaled * 10) * X_orig + (bias_scaled * 100)
w_original_scale = w * (100.0 / 10.0)
bias_original_scale = bias * 100.0

print(f'  Original scale w = {w_original_scale:.6f}')
print(f'  Original scale bias = {bias_original_scale:.6f}')

print(f'\nModel: Score = {w_original_scale:.2f} * Hours + {bias_original_scale:.2f}')
````

````python
#Part C
#Hours to predict for
new_hours = 9

# Use the original scale parameters to make the prediction
predicted_score = w_original_scale * new_hours + bias_original_scale

print(f'Predicting score for {new_hours} hours of study:')
print(f'  Model: Score = {w_original_scale:.2f} * Hours + {bias_original_scale:.2f}')
print(f'  Predicted Score = {w_original_scale:.2f} * {new_hours} + {bias_original_scale:.2f}')
print(f'  Predicted Score = {predicted_score:.2f}')
````

````python
#Part D: Calculate R²

# Calculate final predictions using the scaled parameters
y_final_scaled = w * X_scaled + bias

# Unscale y_final_scaled back to original scale for R^2 calculation with original y
final_predictions_original_scale = y_final_scaled * 100.0

# Mean of the actual y values
y_mean = np.mean(y)

# Total sum of squares (SS_tot)
# Measures the total variance in the actual y values
ss_tot = np.sum((y - y_mean) ** 2)

# Sum of squared residuals (SS_res)
# Measures the variance that the model *could not* explain
ss_res = np.sum((y - final_predictions_original_scale) ** 2)

# Calculate R-squared
r_squared = 1 - (ss_res / ss_tot)

print('R² Score Calculation:')
print(f'  Mean of actual scores (y_mean) = {y_mean:.2f}')
print(f'  SS_res (sum of squared errors) = {ss_res:.4f}')
print(f'  SS_tot (total sum of squares)  = {ss_tot:.4f}')
print(f'\n  R² = 1 - ({ss_res:.4f} / {ss_tot:.4f}) = {r_squared:.6f}')
print(f'\n  The R² score is {r_squared:.6f}, which means our model explains {r_squared*100:.2f}% of the variance in scores.')
````

## Exercise 3: Multiple Features

Add a second feature (number of bedrooms). Your model becomes:

$$\text{price} = w_1 \times \text{size} + w_2 \times \text{bedrooms} + \text{bias}$$

````python
# ====== STEP 1: Create Feature Matrix (X) and Target Vector (y) ======

# Feature matrix X — what we know (house sizes)
X = np.array([600, 800, 1000, 1200, 1400, 1600])  # 6 houses, 1 feature each

# Target vector y — what we want to predict (prices)
y = np.array([120000, 160000, 200000, 240000, 280000, 320000])

print('Feature Matrix X (house sizes):')
print(f'  X = {X}')
print(f'  Shape: {X.shape} — {len(X)} houses, 1 feature\n')

print('Target Vector y (prices):')
print(f'  y = {y}')
print(f'  Shape: {y.shape} — {len(y)} prices\n')

print('Each row in X corresponds to a row in y:')
for i in range(len(X)):
    print(f'  House {i+1}: {X[i]:,} sq ft → £{y[i]:,}')
````

````python
# ======================================================================
# LINEAR REGRESSION FROM SCRATCH — Complete Implementation (Multi-feature)
# ======================================================================

import numpy as np

# ====== STEP 1: Our Data (Linear Algebra) ======
# X now contains two features: [size, bedrooms]
X = np.array([[600, 2],
              [800, 3],
              [1000, 3],
              [1200, 4],
              [1400, 4],
              [1600, 5]])   # sizes and bedrooms
y = np.array([120000, 160000, 200000, 240000, 280000, 320000])  # prices

# Scale data for stable training
# Scale sizes by 1000, bedrooms by 5, and prices by 100000
X_scaled = X.copy().astype(float) # Create a float copy
X_scaled[:, 0] = X_scaled[:, 0] / 1000.0 # Scale sizes
X_scaled[:, 1] = X_scaled[:, 1] / 5.0    # Scale bedrooms
y_scaled = y / 100000.0

print('Feature Matrix X (sizes, bedrooms):')
print(f'  X =\n{X}')
print(f'  Shape: {X.shape} — {len(X)} houses, {X.shape[1]} features\n')

print('Target Vector y (prices):')
print(f'  y = {y}')
print(f'  Shape: {y.shape} — {len(y)} prices\n')

print('Scaled Feature Matrix X_scaled:')
print(f'  X_scaled =\n{X_scaled}')
print(f'Scaled Target Vector y_scaled:  y_scaled = {y_scaled}\n')

# ====== STEP 2: Initialise Parameters ======
w = np.array([0.0, 0.0]) # w is now a vector [w1, w2]
bias = 0.0               # bias (random start)
lr = 0.1                 # learning rate
n = len(X_scaled)        # Number of samples
epochs = 100             # Number of training iterations

# ====== STEP 3: Training Loop (Calculus) ======
print('Training Linear Regression with two features from scratch...')
print(f'{"Epoch":>6} | {"w1":>10} | {"w2":>10} | {"bias":>10} | {"MSE":>14}')
print('-' * 60)

for epoch in range(epochs):
    # Forward pass: predict
    # y_pred = w1 * X_scaled[:,0] + w2 * X_scaled[:,1] + bias
    y_pred = np.dot(X_scaled, w) + bias

    # Compute loss (MSE)
    mse = np.mean((y_pred - y_scaled) ** 2)

    # Compute gradients (partial derivatives)
    dw = (2/n) * np.dot((y_pred - y_scaled), X_scaled) # dw will be a vector [dw1, dw2]
    db = (2/n) * np.sum(y_pred - y_scaled)

    # Update parameters (gradient descent)
    w = w - lr * dw
    bias = bias - lr * db

    if epoch % 10 == 0 or epoch == epochs - 1:
        print(f'{epoch:>6} | {w[0]:>10.4f} | {w[1]:>10.4f} | {bias:>10.4f} | {mse:>14.6f}')

print('-' * 60)

print(f'\nFinal parameters after {epochs} epochs:')
print(f'  Scaled w1 (size) = {w[0]:.6f}')
print(f'  Scaled w2 (bedrooms) = {w[1]:.6f}')
print(f'  Scaled bias = {bias:.6f}')

# Unscaling the parameters to interpret them in the original units
w1_original_scale = w[0] * (100000.0 / 1000.0) # price_scale / size_scale
w2_original_scale = w[1] * (100000.0 / 5.0)   # price_scale / bedrooms_scale
bias_original_scale = bias * 100000.0

print(f'  Original scale w1 (size) = {w1_original_scale:.6f}')
print(f'  Original scale w2 (bedrooms) = {w2_original_scale:.6f}')
print(f'  Original scale bias = {bias_original_scale:.6f}')

print(f'\nModel: Price = {w1_original_scale:.2f} * Size + {w2_original_scale:.2f} * Bedrooms + {bias_original_scale:.2f}')

# ====== STEP 4: Make a Prediction ======
# For a house of 1100 sq ft with 4 bedrooms
new_features_original = np.array([1100, 4])
# Scale the new features using the same scaling factors as training data
new_features_scaled = np.array([new_features_original[0] / 1000.0, new_features_original[1] / 5.0])

# Predict using the scaled model and then unscale the output
predicted_scaled = np.dot(new_features_scaled, w) + bias
predicted_original = predicted_scaled * 100000.0

print(f'\n Predicted price for {new_features_original[0]} sq ft and {new_features_original[1]} bedrooms: £{predicted_original:,.0f}')

# ====== STEP 5: Evaluate (Statistics) ======
y_final_scaled = np.dot(X_scaled, w) + bias

# Unscale y_final_scaled back to original scale for R^2 calculation with original y
final_predictions_original_scale = y_final_scaled * 100000.0

# Mean of the actual y values
y_mean = np.mean(y)

# Total sum of squares (SS_tot)
ss_tot = np.sum((y - y_mean) ** 2)

# Sum of squared residuals (SS_res)
ss_res = np.sum((y - final_predictions_original_scale) ** 2)

# Calculate R-squared
r_squared = 1 - (ss_res / ss_tot)

print(f' R² Score: {r_squared:.6f}')
````

````python
# ====== SCIKIT-LEARN VERSION ======

from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X.reshape(-1, 1), y)              # Train
prediction = model.predict([[1100]])          # Predict
score = model.score(X.reshape(-1, 1), y)     # Evaluate

print(f'Scikit-Learn Results:')
print(f'  w (coefficient) = {model.coef_[0]:.1f}')
print(f'  bias (intercept) = {model.intercept_:.1f}')
print(f'  Prediction for 1100 sq ft = £{prediction[0]:,.0f}')
print(f'  R² Score = {score:.6f}')
print(f'\n Same answer! But now you understand WHAT those 4 lines actually DO.')
````
